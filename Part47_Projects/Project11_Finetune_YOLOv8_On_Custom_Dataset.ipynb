# Project 11 — Fine-Tune YOLOv8 on a Custom Dataset

Build a complete custom object detection system. Annotate 200+ images using CVAT or LabelImg, prepare the dataset in YOLO format, train YOLOv8 for 50 epochs, evaluate using mAP@0.5, export the trained model to ONNX, and deploy it on a live webcam feed.
